# Model 1 — Detecção de Veículo

Este notebook treina o **primeiro modelo YOLO** do pipeline ANPR.

**Função:** Recebe uma imagem de cena completa e retorna onde está o carro.

**Pipeline completo:**
```
Câmera → [Model 1: detecta carro] → [Model 2: detecta placa no carro] → [OCR: lê o texto]
```

**Dataset:** `dataset_vehicle` — gerado pelo script `generate_vehicle_dataset.py`
- Train: 236 imagens
- Valid: 68 imagens
- Test: 32 imagens
- Classe: `vehicle`

````python
import os
from ultralytics import YOLO

# Modelo pré-treinado (mesmo usado pelo Gustavo)
# Troque para 'yolov8n.pt' se não tiver o yolo26n.pt disponível
model = YOLO('yolo26n.pt')

config_path = '../datasets/dataset_vehicle/data.yaml'

print(f'Dataset: {config_path}')
print(f'Modelo base: {model.model_name}')
````

````python
# Treino do Model 1
results = model.train(
    data=config_path,
    epochs=200,
    imgsz=640,
    batch=32,
    patience=25,       # para cedo se não melhorar
    plots=True,
    name='model-vehicle'
)
````

````python
# Avaliação no conjunto de validação
metrics = model.val(data=config_path)

print(f"mAP50:    {metrics.box.map50:.4f}")
print(f"mAP50-95: {metrics.box.map:.4f}")
print(f"Precision:{metrics.box.mp:.4f}")
print(f"Recall:   {metrics.box.mr:.4f}")
````

````python
# Localização do melhor modelo salvo
best_model_path = results.save_dir + '/weights/best.pt'
print(f'Melhor modelo salvo em: {best_model_path}')
````
